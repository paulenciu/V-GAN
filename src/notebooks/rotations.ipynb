```python
import torch
import torch.nn.functional as F

# Example batch of images (n, 3, 32, 32)
n, c, h, w = 8, 3, 32, 32  # Example dimensions
images = torch.randn(n, c, h, w)  # Random batch of images

# Define the 2x2 rotation matrix
theta = torch.tensor([[0.866, -0.5],  # Example rotation matrix for 30 degrees
                      [0.5, 0.866]], dtype=torch.float32)

# Extend the rotation matrix to 3x3 to include translation terms
# Add the third row for homogeneous coordinates
transformation_matrix = torch.eye(3)  # 3x3 Identity matrix
transformation_matrix[:2, :2] = theta  # Insert the 2x2 rotation matrix

# Repeat the transformation matrix for the batch
transformation_matrices = transformation_matrix.unsqueeze(0).repeat(n, 1, 1)  # Shape (n, 3, 3)

# Normalize the grid to [-1, 1] for grid_sample
grid = F.affine_grid(transformation_matrices[:, :2, :], images.size(), align_corners=False)

# Apply the transformation using grid_sample
rotated_images = F.grid_sample(images, grid, align_corners=False)

# Verify the output shape
print("Rotated Images Shape:", rotated_images.shape)
```
