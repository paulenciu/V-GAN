```python
import torch
import torchvision
import torchvision.transforms as transforms

from src.models.autoencoder.resnet.RestNetAutoEncoder import ResNet50AutoEncoder
```

```python
dataset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transforms.ToTensor())
cats_dataset = [(img, label) for (img, label) in dataset if label == 3]
cats_dataloader = torch.utils.data.DataLoader(cats_dataset, batch_size=1, shuffle=True)
```

```python
device = torch.device('cuda:0' if torch.cuda.is_available() else 'mps:0' if torch.backends.mps.is_available() else 'cpu')
model = ResNet50AutoEncoder()
encoder = model.model.encoder.to(device)
```

```python
# Fetch one image and its label
image, label = next(iter(cats_dataloader))
print(f"Image shape: {image.shape}")
print(f"Label: {label}")
encoder.eval()
output = encoder(image.to(device))
output.shape
```
